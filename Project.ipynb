```python
%%writefile app.py
import streamlit as slt
from sklearn.linear_model import LinearRegression
import pandas as pd
data={
     'area':[1000,1107,1165,1200,1256,1287,1500,1535,1568,1800,1875,1824],
      'age':[20,19,17,15,14,12,11,9,7,6,4,2],
      'bedrooms':[2,3,2,4,4,5,5,6,6,2,3,3],
      'price':[40,50,65,80,100,120,125,140,146,67,85,90]
}
df=pd.DataFrame(data)
x=df[['area','age','bedrooms']]
y=df['price']
model=LinearRegression()
model.fit(x,y)
slt.title('house price prediction ')
area=slt.number_input('area-sqft')
age=slt.number_input('age of the house')
bedroom=slt.number_input('total bedroom ')
if slt.button('Predict'):
  prediction=model.predict([[area,age, bedroom]])
  slt.write('cost to buy', prediction[0])
```
