```python
import pandas as pd
from sklearn.linear_model import LinearRegression
#creation of data set
data={
'experience':[1,2,3,4,5],
'salary':[3,5,7,9,11]

}
df=pd.DataFrame(data)
x=df[['experience']]
y=df[['salary']]

model=LinearRegression()
model.fit(x,y)
prediction=model.predict([[6]])
print('expected salary', prediction[0])
```

```python
#multi linear regression

#y=B0+B1X1+B2X2+...+BkXk
#house price
import pandas as pd
from sklearn.linear_model import LinearRegression

#creation of data frame
data={
     'area':[1000,1200,1500,1800],
      'age':[10,8,5,4],
      'bedrooms':[2,3,2,4],
      'price':[40,50,65,80]
}
df=pd.DataFrame(data)
#independent
x=df[['area','age','bedrooms']]
#dependent
y=df['price']

model=LinearRegression()
model.fit(x,y)
prediction=model.predict([[1600,5,3]])
print('cost to buy', prediction[0])
```
