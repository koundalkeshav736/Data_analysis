```python
#multi linear regression

#y=B0+B1X1+B2X2+...+BkXk
#house price
import pandas as pd
from sklearn.linear_model import LinearRegression

#creation of data frame
data={
     'area':[1000,1200,1500,1800],
      'age':[10,8,5,4],
      'bedrooms':[2,3,2,4],
      'price':[40,50,65,80]
}
df=pd.DataFrame(data)
#independent
x=df[['area','age','bedrooms']]
#dependent
y=df['price']

model=LinearRegression()
model.fit(x,y)
prediction=model.predict([[1600,5,3]])
print('cost to buy', prediction[0])
```

```python
#MEAN ABSOLUTE ERROR
from sklearn.metrics import mean_absolute_error

y_true=[3,5,7,9,11]
y_pred=[2,5,6,7,10]
data=mean_absolute_error(y_true,y_pred)
print('output',data)
```

```python
#mean squared error
from sklearn.metrics import mean_squared_error

y_true=[3,5,7,9,11]
y_pred=[2,5,6,7,10]
data=mean_squared_error(y_true,y_pred)
print('output',data)
```

```python
#root mean square error
```

```python

from sklearn.metrics import root_mean_squared_error

y_true=[3,5,7,9,11]
y_pred=[2,5,6,7,10]
data=root_mean_squared_error(y_true,y_pred)
print('output',data)
```

```python
#r^2 score method

from sklearn.metrics import r2_score
y_true=[3,5,7,9,11]
y_pred=[2,5,6,7,10]
r2=r2_score(y_true,y_pred)
print('r2_score',r2)
```

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
```

```python
data={
    'salary':[20000,30000,40000,60000,70000,90000],
    'approved':[0,0,0,1,1,1]
}
df=pd.DataFrame(data)
x=df[['salary']].values
y=df['approved'].values.ravel()

model=LogisticRegression()
model.fit(x,y)
prediction=model.predict([[65000]])
print(prediction)
```

```python
#decision tree
from sklearn.tree import DecisionTreeClassifier
model=DecisionTreeClassifier()
model.fit(x,y)
pre=model.predict([[65000]])
print(pre)
```
