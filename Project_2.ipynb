```python
cate_val=[]
cont_val=[]
for column in df.columns:
  if df[column].nunique()<=10:
    cate_val.append(column)
  else:
    cont_val.append(column)
    print(cate_val)
```

```python
from sklearn.preprocessing import StandardScaler
st=StandardScaler()
df[cont_val]=st.fit_transform(df[cont_val])
```

```python
df.head()
```

```python
#traing and testing data
```

```python
x=df.drop('target',axis=1)
y=df['target']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

logistic regression

```python
#logistic regression
from sklearn.linear_model import LogisticRegression
```

```python
lr=LogisticRegression()
lr.fit(x_train,y_train)
```

```python
y_predict=lr.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
x1=accuracy_score(y_test,y_predict)
print(x1)
```

```python
#support vector classifier(svc)
from sklearn.svm import SVC
svm=SVC()
svm.fit(x_train,y_train)
```

```python
y_predict2=svm.predict(x_test)
x2=accuracy_score(y_test,y_predict2)
print(x2)
```

```python
#knn
from sklearn.neighbors import KNeighborsClassifier
knn=KNeighborsClassifier()
knn.fit(x_train,y_train)
```

```python
y_predict3=knn.predict(x_test)
x3=accuracy_score(y_test,y_predict3)
print(x3)
```

```python
#decission tree
from sklearn.tree import DecisionTreeClassifier
dt=DecisionTreeClassifier()
dt.fit(x_train,y_train)
```

```python
y_predict4=dt.predict(x_test)
x4=accuracy_score(y_test,y_predict4)
print(x4)
```

```python
#randomForest Classifier
from sklearn.ensemble import RandomForestClassifier
rf=RandomForestClassifier()
rf.fit(x_train,y_train)
```

```python
y_predict5=rf.predict(x_test)
x5=accuracy_score(y_test,y_predict5)
print(x5)
```

```python
#gradient boosting classifier
from sklearn.ensemble import GradientBoostingClassifier
gb=GradientBoostingClassifier()
gb.fit(x_train,y_train)
```

```python
y_predict6=gb.predict(x_test)
x6=accuracy_score(y_test,y_predict6)
print(x6)
```

```python
print("logistic regression",x1)
print("support vector classifier",x2)
print("knn",x3)
print("decission tree",x4)
print("randomForest Classifier",x5)
print("gradient boosting classifier",x6)
```
